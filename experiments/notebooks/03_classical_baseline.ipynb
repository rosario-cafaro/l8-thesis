# 03 - Baseline classiche

Addestra i tre modelli classici di riferimento (Logistic Regression, SVM a kernel RBF, Random Forest) sulle feature preprocessate salvate dal notebook 02, ottimizzando gli iperparametri tramite grid search con validazione incrociata stratificata a 5 fold (sezione "Modelli classici di baseline" del Capitolo 7).

```python
import sys
from pathlib import Path

sys.path.append(str(Path.cwd().parent))

import numpy as np
import pandas as pd

from config import DATASET_NAMES, RANDOM_STATE, CV_FOLDS
from src.pipeline import train_classical_models

PROCESSED_DIR = Path.cwd().parent / "data" / "processed"
TABLES_DIR = Path.cwd().parent / "results" / "tables"
TABLES_DIR.mkdir(parents=True, exist_ok=True)
```

```python
all_rows = []

for name in DATASET_NAMES:
    data = np.load(PROCESSED_DIR / f"{name}.npz")
    rows = train_classical_models(
        data["X_train"], data["y_train"], data["X_test"], data["y_test"],
        random_state=RANDOM_STATE, cv_folds=CV_FOLDS,
    )
    for row in rows:
        row["dataset"] = name
    all_rows.extend(rows)

classical_results = pd.DataFrame(all_rows)
classical_results.to_csv(TABLES_DIR / "classical_baseline_results.csv", index=False)
classical_results[["dataset", "model", "accuracy", "precision", "recall", "f1_score", "fit_time_s"]]
```
