# 01 - Acquisizione e validazione dei dati

Carica i tre dataset selezionati (Breast Cancer Wisconsin, Wine, Digits) tramite Scikit-learn, esegue la validazione preliminare descritta nella sezione "Acquisizione e validazione dei dati" (Capitolo 7) e salva una copia in `data/raw/` per le fasi successive della pipeline.

```python
import sys
from pathlib import Path

sys.path.append(str(Path.cwd().parent))

from src.data.acquisition import load_dataset

DATASETS = ["breast_cancer", "wine", "digits"]
RAW_DIR = Path.cwd().parent / "data" / "raw"
RAW_DIR.mkdir(parents=True, exist_ok=True)
```

```python
summary_rows = []

for name in DATASETS:
    X, y = load_dataset(name)
    X.assign(target=y).to_csv(RAW_DIR / f"{name}.csv", index=False)
    summary_rows.append({
        "dataset": name,
        "n_samples": X.shape[0],
        "n_features": X.shape[1],
        "n_classes": y.nunique(),
        "missing_values": int(X.isnull().sum().sum()),
    })

import pandas as pd
summary = pd.DataFrame(summary_rows)
summary
```

La tabella riprodotta sopra corrisponde alla Tabella `validazione_dataset` del Capitolo 7: i tre dataset non presentano valori mancanti e sono composti esclusivamente da feature numeriche.
